### Часть 3. Базовая графика с `matplotlib`

Простейший способ визуализировать форму распределения — построить гистограмму.

```python
plt.hist(df["body_mass_kg"], bins=25)
plt.show()
```

Для сравнения двух распределений удобно задать прозрачность через параметр `alpha`.

```python
plt.hist(df["bill_depth_z"], bins=20)
plt.hist(df["flipper_length_z"], bins=20, alpha=0.75)
plt.show()
```

Для сравнения множества распределений удобно использовать диаграммы размаха, в простонародье "ящики с усами".

```python
orig_numeric_cols = [
    "body_mass_kg",
    "bill_length_mm",
    "bill_depth_mm",
    "flipper_length_mm",
]
plt.boxplot(df[orig_numeric_cols], tick_labels=orig_numeric_cols)
plt.show()
```

Зависимость двух переменных обычно оценивают с диаграммой рассеяния.

```python
plt.scatter(df["body_mass_kg"], df["flipper_length_mm"])
plt.show()
```

`plt.subplots` позволяет собирать множество графиков на сетке.

```python
fig, axes = plt.subplots(2, 2)

ax = axes[0, 0]
ax.hist(df["body_mass_kg"], edgecolor="white")

ax = axes[0, 1]
ax.scatter(df["flipper_length_mm"], df["body_mass_kg"], alpha=0.5)

ax = axes[1, 0]
ax.scatter(df["body_mass_kg"], df["flipper_length_mm"], alpha=0.5)

ax = axes[1, 1]
ax.hist(df["flipper_length_mm"], edgecolor="white")

plt.tight_layout()
plt.show()
```
